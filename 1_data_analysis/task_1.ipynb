```python
from numpy import astype
import pandas as pd

url = "https://pstu.ru/files/file/Abitur/2025%20final/%D0%9C%D0%A2%D0%A4_%D0%91_%D0%9D_000007048.html"

# загрузка таблицы
df = pd.read_html(url)[0]

# удаление пустых строк
df = df.dropna(how="all")

# удаление ненужного текста
df = df.drop(df.index[:10])

# удаление лтшних столбцов
df = df.drop(columns=[0, 1, 7, 8, 9, 10, 11, 12])

# переименование индексов
df = df.reset_index(drop=True)


# пункт 4
headers = df.iloc[0]         # первая строка дата фрейма
df = df[1:]                  # удаление первой строки из дата фрейма
df.columns = headers         # переименование названий столбцов

# приведение к int
for col in df.columns:
    try:
        df[col] = pd.to_numeric(df[col].astype(str), errors='coerce').astype('Int64')
    except Exception:
        pass

print(df.dtypes)
display(df)


# пункт 5
# определение среднего, минимального и максимального суммарных баллов
mean_score = df["Сумма баллов"].mean()
min_score = df['Сумма баллов'].min()
max_score = df['Сумма баллов'].max()


print(f"Средний суммарный балл: {mean_score:.2f}")
print(f"Минимальный суммарный балл: {min_score}")
print(f"Максимальный суммарный балл: {max_score}")


# средние значения по трем сдаваемым предметам ЕГЭ
subjects = ["АиНМА / Математика", "ПФ / Информатика и ИКТ / Физика", "Русский язык"]

# таблтица со средними значениями по предметам
df_mean = pd.DataFrame({
    "Предмет": subjects,
    "Средний балл": [df[subj].mean() for subj in subjects]
})

# предмет с максимальным средним баллом
max_subject = df_mean.loc[df_mean["Средний балл"].idxmax(), "Предмет"]

display(df_mean)
print(f"Предмет, по которому абитуриенты в среднем набирают больше баллов: {max_subject}")


# пункт 6
# кол-во абитуриентов, у которых баллы по русскому ниже ср балла по русскому, а по остальным предметам выше ср
mean_math = df_mean.loc[df_mean["Предмет"] == "АиНМА / Математика", "Средний балл"].values[0]
mean_phys_inf = df_mean.loc[df_mean["Предмет"] == "ПФ / Информатика и ИКТ / Физика", "Средний балл"].values[0]
mean_rus = df_mean.loc[df_mean["Предмет"] == "Русский язык", "Средний балл"].values[0]

# логическое условие для Тип 1 (Tru/False)
type_1_conditions = (
    (df["Русский язык"] < mean_rus) &
    (df["АиНМА / Математика"] > mean_math) &
    (df["ПФ / Информатика и ИКТ / Физика"] > mean_phys_inf)
)

# фильтр по соответствию условию
type_1_students = df[type_1_conditions]

# подсчет кол-ва абитуриентов
type_1_count = len(type_1_students)

print(f"\nКоличество абитуриентов Типа 1: {type_1_count}")


# пункт 7
# кол-во абитуриентов, у которых балл по русскому выше ср балла по русскому языку, а по остальным предметам ниже ср
mean_math = df_mean.loc[df_mean["Предмет"] == "АиНМА / Математика", "Средний балл"].values[0]
mean_phys_inf = df_mean.loc[df_mean["Предмет"] == "ПФ / Информатика и ИКТ / Физика", "Средний балл"].values[0]
mean_rus = df_mean.loc[df_mean["Предмет"] == "Русский язык", "Средний балл"].values[0]

# логическое условие для Тип 2 (Tru/False)
type_2_conditions = (
    (df["Русский язык"] > mean_rus) &
    (df["АиНМА / Математика"] < mean_math) &
    (df["ПФ / Информатика и ИКТ / Физика"] < mean_phys_inf)
)

# фильтр по соответствию условию
type_2_students = df[type_2_conditions]

# подсчет кол-ва абитуриентов
type_2_count = len(type_2_students)

print(f"\nКоличество абитуриентов Типа 2: {type_2_count}")


# пункт 8
# создание и заполнение нового DataFrame
df_new = pd.DataFrame({
    "Подразделение": ["Механико-технологический факультет"],
    "Уровень подготовки": ["Бакалавриат"], 
    "Направление подготовки/специальность": [" Управление качеством"], 
    "Год": [2025],
    "Количество мест": [len(df)],
    "Предметы ЕГЭ": [", ".join(["АиНМА / Математика", "ПФ / Информатика и ИКТ / Физика", "Русский язык"])],
    "Средняя сумма баллов": [round(mean_score, 3)],
    "Min сумма баллов": [min_score],
    "Max сумма баллов": [max_score],
    "Предмет с высшим средним": [max_subject],
    "Кол-во студентов тип_1": [type_1_count],
    "Кол-во студентов тип_2": [type_2_count]
})

display(df_new)
```
